# Bandedge offset and Fermi level pinning

This notebook is used to explore the calculation of the Fermi level across the device. The relative band offsets between different layers (Ge/SiGe) are tweaked in order to set the Fermi level to zero, but not constant across the whole device. This is expected to fix the screening issue that prevents accumulation in the QW due to fast accumulation in the SiGe cap.

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from matplotlib.colors import ListedColormap, BoundaryNorm
from matplotlib.patches import Rectangle

def _find_unique(folder, patterns):
    folder = Path(folder)
    matches = []
    for pat in patterns:
        matches.extend(folder.glob(pat))
    matches = sorted(set(matches))
    if not matches:
        raise FileNotFoundError(
            f"No file found in {folder} matching any of: {patterns}"
        )
    if len(matches) > 1:
        print("Multiple matches found, using:", matches[0].name)
        print("All matches:", [m.name for m in matches])
    return matches[0]

def section_fld(folder, stem, section_name):
    return _find_unique(
        folder,
        [
            f"{stem}_{section_name}.fld",
            f"{stem}_2d_{section_name}.fld",
            f"{stem}*{section_name}*.fld",
        ],
    )

def section_dat(folder, stem, section_name):
    return _find_unique(
        folder,
        [
            f"{stem}_{section_name}.dat",
            f"{stem}_1d_{section_name}.dat",
            f"{stem}*{section_name}*.dat",
        ],
    )

def first_col(df):
    return df.columns[0]

def last_col(df):
    return df.columns[-1]

def print_matching_files(folder, contains=""):
    files = sorted([p.name for p in Path(folder).iterdir() if contains in p.name])
    for f in files:
        print(f)
```

```python
def read_nextnano_fld_rectilinear(path: Path):
    """
    Read nextnano AVS/Express .fld (ASCII, rectilinear).
    Returns: x_coords, y_coords, data (shape: [ny, nx] for scalar; [ny, nx, veclen] for vector)
    """
    path = Path(path)
    text = path.read_text(errors="ignore").splitlines()

    def get_int(key):
        m = re.search(rf"^{key}\s*=\s*(\d+)\s*$", "\n".join(text), flags=re.MULTILINE)
        if not m:
            raise ValueError(f"Could not find '{key} =' in header")
        return int(m.group(1))

    def get_str(key):
        m = re.search(rf"^{key}\s*=\s*(.+)\s*$", "\n".join(text), flags=re.MULTILINE)
        if not m:
            raise ValueError(f"Could not find '{key} =' in header")
        return m.group(1).strip()

    ndim = get_int("ndim")
    if ndim != 2:
        raise ValueError(f"Only ndim=2 supported here, got ndim={ndim}")

    nx = get_int("dim1")
    ny = get_int("dim2")
    veclen = get_int("veclen")
    field_type = get_str("field")
    if field_type != "rectilinear":
        raise ValueError(f"Only rectilinear supported, got field={field_type}")

    # parse skip lines for coords and data from the header lines like:
    # coord 1 ... skip=15
    # coord 2 ... skip=83
    # variable 1 ... skip=638
    header_blob = "\n".join(text)

    def get_skip(kind, idx):
        m = re.search(rf"^{kind}\s+{idx}\s+.*?\bskip=(\d+)\b", header_blob, flags=re.MULTILINE)
        if not m:
            raise ValueError(f"Could not find '{kind} {idx} ... skip=' in header")
        return int(m.group(1))

    skip_x = get_skip("coord", 1)
    skip_y = get_skip("coord", 2)
    skip_data = get_skip("variable", 1)

    # Helper to read a block of floats from lines starting at skip
    def read_floats_from_lines(start_line, count):
        vals = []
        i = start_line
        while i < len(text) and len(vals) < count:
            line = text[i].strip()
            if line and not line.startswith("#"):
                # some files may have multiple numbers per line
                vals.extend([float(v) for v in line.split()])
            i += 1
        if len(vals) < count:
            raise ValueError(f"Expected {count} floats from line {start_line}, got {len(vals)}")
        return np.array(vals[:count], dtype=float)

    x = read_floats_from_lines(skip_x, nx)
    y = read_floats_from_lines(skip_y, ny)

    data_count = nx * ny * veclen
    raw = read_floats_from_lines(skip_data, data_count)

    if veclen == 1:
        Z = raw.reshape((ny, nx))  # [y, x]
        return x, y, Z
    else:
        Z = raw.reshape((ny, nx, veclen))
        return x, y, Z
```

#### - Change Ge bandoffset parameter

```python
from pathlib import Path
from datetime import datetime
import re
import shutil

def set_ge_valence_bandoffset(
    new_value,
    database_path="/Applications/nextnano/2025_12_17/nextnano++/database/database.nnp",
    make_backup=True,
):
    path = Path(database_path).expanduser()
    new_value_str = (
        f"{new_value:.15g}"
        if isinstance(new_value, (int, float))
        else str(new_value).strip()
    )

    # nextnano's database file here is not UTF-8; latin-1 preserves it safely.
    text = path.read_text(encoding="latin-1")
    lines = text.splitlines(keepends=True)

    stack = []
    target_binary_depth = None
    old_value = None
    line_number = None

    def brace_events(code):
        token = ""
        for ch in code:
            if ch.isalnum() or ch == "_":
                token += ch
            elif ch == "{":
                yield ("open", token.strip() or None)
                token = ""
            elif ch == "}":
                yield ("close", None)
                token = ""
            else:
                token = ""

    for idx, line in enumerate(lines):
        code = line.split("#", 1)[0]

        if target_binary_depth is None and re.fullmatch(r"\s*name\s*=\s*Ge\s*", code):
            if stack and stack[-1] == "binary_zb":
                target_binary_depth = len(stack)

        in_target_ge = (
            target_binary_depth is not None
            and len(stack) >= target_binary_depth
            and stack[target_binary_depth - 1] == "binary_zb"
        )
        in_ge_valence = in_target_ge and "valence_bands" in stack[target_binary_depth:]

        if in_ge_valence and re.match(r"^\s*bandoffset\s*=", code):
            match = re.match(
                r"^(\s*bandoffset\s*=\s*)([-+]?(?:\d+(?:\.\d*)?|\.\d+)(?:[eE][-+]?\d+)?)(\s*(?:#.*)?(?:\r?\n)?)$",
                line,
            )
            if not match:
                raise RuntimeError(f"Could not parse bandoffset line: {line!r}")

            old_value = match.group(2)
            lines[idx] = f"{match.group(1)}{new_value_str}{match.group(3)}"
            line_number = idx + 1
            break

        for action, name in brace_events(code):
            if action == "open":
                stack.append(name)
            else:
                if not stack:
                    raise RuntimeError("Unbalanced braces while parsing database.nnp")
                stack.pop()
                if target_binary_depth is not None and len(stack) < target_binary_depth:
                    target_binary_depth = None

    if old_value is None:
        raise RuntimeError("Could not find Ge -> valence_bands -> bandoffset in database.nnp")

    if make_backup:
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
        backup_path = path.with_name(f"{path.name}.bak_{timestamp}")
        shutil.copy2(path, backup_path)
    else:
        backup_path = None

    path.write_text("".join(lines), encoding="latin-1")

    return {
        "path": str(path),
        "line": line_number,
        "old_value": old_value,
        "new_value": new_value_str,
        "backup": str(backup_path) if backup_path else None,
    }
```

```python
from pathlib import Path
import io
import re
import time

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
from PIL import Image as PILImage
from IPython.display import Image as IPyImage, display

required_names = ["REPO_ROOT", "INPUT_DOUBLE_DOT_2D", "nna", "set_ge_valence_bandoffset"]
missing = [name for name in required_names if name not in globals()]
if missing:
    raise RuntimeError("Run the setup cells first. Missing: " + ", ".join(missing))

DATABASE_NNP = Path("/Applications/nextnano/2025_12_17/nextnano++/database/database.nnp")
BANDOFFSET_VALUES = np.round(np.arange(0.0, 2.0 + 0.001, 0.1), 1)
GIF_DIR = REPO_ROOT / "notebooks" / "analysis" / "artifacts" / "bandoffset_sweep_2d"
GIF_DIR.mkdir(parents=True, exist_ok=True)

FRAME_DURATION_MS = 350
FRAME_DPI = 110
SHOW_INLINE = True  # set False if you do not want the GIFs displayed in the notebook


def brace_events(code):
    token = ""
    for ch in code:
        if ch.isalnum() or ch == "_":
            token += ch
        elif ch == "{":
            yield ("open", token.strip() or None)
            token = ""
        elif ch == "}":
            yield ("close", None)
            token = ""
        else:
            token = ""


def get_ge_valence_bandoffset(database_path=DATABASE_NNP):
    path = Path(database_path).expanduser()
    lines = path.read_text(encoding="latin-1").splitlines()

    stack = []
    target_binary_depth = None

    for line in lines:
        code = line.split("#", 1)[0]

        if target_binary_depth is None and re.fullmatch(r"\s*name\s*=\s*Ge\s*", code):
            if stack and stack[-1] == "binary_zb":
                target_binary_depth = len(stack)

        in_target_ge = (
            target_binary_depth is not None
            and len(stack) >= target_binary_depth
            and stack[target_binary_depth - 1] == "binary_zb"
        )
        in_ge_valence = in_target_ge and "valence_bands" in stack[target_binary_depth:]

        if in_ge_valence:
            match = re.match(
                r"^\s*bandoffset\s*=\s*([-+]?(?:\d+(?:\.\d*)?|\.\d+)(?:[eE][-+]?\d+)?)\s*(?:#.*)?$",
                line,
            )
            if match:
                return float(match.group(1))

        for action, name in brace_events(code):
            if action == "open":
                stack.append(name)
            else:
                if not stack:
                    raise RuntimeError("Unbalanced braces while parsing database.nnp")
                stack.pop()
                if target_binary_depth is not None and len(stack) < target_binary_depth:
                    target_binary_depth = None

    raise RuntimeError("Could not find Ge -> valence_bands -> bandoffset in database.nnp")


def fig_to_pil(fig, dpi=FRAME_DPI):
    buf = io.BytesIO()
    fig.savefig(buf, format="png", dpi=dpi, bbox_inches="tight", facecolor="white")
    plt.close(fig)
    buf.seek(0)
    return PILImage.open(buf).convert("RGB").copy()


def save_gif(frames, path, duration_ms=FRAME_DURATION_MS):
    path = Path(path)
    if not frames:
        raise ValueError(f"No frames available for {path.name}")
    frames[0].save(
        path,
        save_all=True,
        append_images=frames[1:],
        duration=duration_ms,
        loop=0,
        disposal=2,
    )
    return path


def padded_limits(values, pad_frac=0.06):
    vals = np.asarray(values, dtype=float)
    vals = vals[np.isfinite(vals)]
    vmin = float(vals.min())
    vmax = float(vals.max())
    span = vmax - vmin
    pad = pad_frac * span if span > 0 else pad_frac * max(abs(vmax), 1.0)
    return vmin - pad, vmax + pad


def load_2d_snapshot(run_root, offset):
    run_root = Path(run_root)
    bias0 = run_root / "bias_00000"
    structure = run_root / "Structure"

    xc, zc, contacts = nna.load_fld_scalar(structure / "contacts.fld")
    contacts = contacts.astype(int)

    contact_names = {1: "BG1", 2: "PG1", 3: "BG2", 4: "PG2", 5: "BG3"}
    rows = []
    for cid in [1, 2, 3, 4, 5]:
        mask = contacts == cid
        iz, ix = np.where(mask)
        rows.append(
            {
                "contact": contact_names[cid],
                "x_left_nm": xc[ix].min(),
                "x_right_nm": xc[ix].max(),
                "x_center_nm": 0.5 * (xc[ix].min() + xc[ix].max()),
            }
        )
    gate_df = pd.DataFrame(rows).sort_values("x_center_nm").reset_index(drop=True)

    xlim = (gate_df["x_left_nm"].min() - 70, gate_df["x_right_nm"].max() + 70)
    zlim = (-40, 180)
    x_window = xlim
    z_window = (-30, 110)

    _, z_mat, materials = nna.load_fld_scalar(structure / "materials.fld")
    materials = materials.astype(int)

    ge_mask = materials == 8
    iz_ge, _ = np.where(ge_mask)
    z_ge_min = z_mat[iz_ge].min()
    z_ge_max = z_mat[iz_ge].max()

    semi_mask = np.isin(materials, [8, 16])
    iz_semi, _ = np.where(semi_mask)
    z_semi_top = z_mat[iz_semi].max()

    x_be, z_be, hh_2d, _ = nna.load_fld_variable(bias0 / "bandedges.fld", 9)
    _, _, lh_2d, _ = nna.load_fld_variable(bias0 / "bandedges.fld", 10)
    band_x = nna.read_dat(bias0 / "bandedges_1d_x_QD.dat")
    band_x = band_x[(band_x["x[nm]"] >= xlim[0]) & (band_x["x[nm]"] <= xlim[1])].copy()

    x_den, z_den, p2d = nna.load_fld_scalar(bias0 / "density_hole.fld")
    p2d_cm3 = p2d * 1e18
    p1d = nna.read_dat(bias0 / "density_hole_1d_x_QD.dat")
    p1d = p1d[(p1d["x[nm]"] >= xlim[0]) & (p1d["x[nm]"] <= xlim[1])].copy()
    cuts_density_z = {
        name: nna.read_dat(bias0 / f"density_hole_1d_z_{name}.dat")
        for name in ["PG1", "BG2", "PG2"]
    }

    x_phi, z_phi, phi2d = nna.load_fld_scalar(bias0 / "potential.fld")
    phi_x = nna.read_dat(bias0 / "potential_1d_x_QD.dat")
    phi_x = phi_x[(phi_x["x[nm]"] >= x_window[0]) & (phi_x["x[nm]"] <= x_window[1])].copy()
    cuts_phi_z = {
        name: nna.read_dat(bias0 / f"potential_1d_z_{name}.dat")
        for name in ["PG1", "BG2", "PG2"]
    }

    return {
        "offset": float(offset),
        "run_root": run_root,
        "gate_df": gate_df,
        "xlim": xlim,
        "zlim": zlim,
        "x_window": x_window,
        "z_window": z_window,
        "z_ge_min": z_ge_min,
        "z_ge_max": z_ge_max,
        "z_semi_top": z_semi_top,
        "x_be": x_be,
        "z_be": z_be,
        "hh_2d": hh_2d,
        "lh_2d": lh_2d,
        "band_x": band_x,
        "x_den": x_den,
        "z_den": z_den,
        "p2d_cm3": p2d_cm3,
        "p1d": p1d,
        "cuts_density_z": cuts_density_z,
        "x_phi": x_phi,
        "z_phi": z_phi,
        "phi2d": phi2d,
        "phi_x": phi_x,
        "cuts_phi_z": cuts_phi_z,
    }


def compute_scales(snapshots):
    band_vals = []
    band_line_vals = []
    phi_map_vals = []
    phi_line_vals = []

    for s in snapshots:
        ixw = (s["x_be"] >= s["xlim"][0]) & (s["x_be"] <= s["xlim"][1])
        izw = (s["z_be"] >= s["zlim"][0]) & (s["z_be"] <= s["zlim"][1])

        band_vals.extend(
            [
                s["hh_2d"][np.ix_(izw, ixw)].ravel(),
                s["lh_2d"][np.ix_(izw, ixw)].ravel(),
            ]
        )
        band_line_vals.extend(
            [
                s["band_x"]["HH[eV]"].to_numpy(),
                s["band_x"]["LH[eV]"].to_numpy(),
                s["band_x"]["hole_Fermi_level[eV]"].to_numpy(),
            ]
        )
        phi_map_vals.append(s["phi2d"].ravel())
        phi_line_vals.append(s["phi_x"][s["phi_x"].columns[-1]].to_numpy())
        phi_line_vals.extend(df[df.columns[-1]].to_numpy() for df in s["cuts_phi_z"].values())

    band_vals = np.concatenate([v[np.isfinite(v)] for v in band_vals])
    phi_map_vals = np.concatenate([v[np.isfinite(v)] for v in phi_map_vals])

    band_vmin, band_vmax = np.percentile(band_vals, [2, 98])
    phi_vmin, phi_vmax = np.percentile(phi_map_vals, [1, 99])

    return {
        "band_vmin": float(band_vmin),
        "band_vmax": float(band_vmax),
        "band_line_ylim": padded_limits(np.concatenate(band_line_vals), pad_frac=0.07),
        "phi_vmin": float(phi_vmin),
        "phi_vmax": float(phi_vmax),
        "phi_line_ylim": padded_limits(np.concatenate(phi_line_vals), pad_frac=0.08),
    }


def annotate_gate_lines(ax, gate_df, y_text, color_line, color_text, bbox_fc=None):
    for row in gate_df.itertuples():
        ax.axvline(row.x_center_nm, color=color_line, ls="--", lw=1, alpha=0.6)
        text_kwargs = dict(ha="center", va="top", fontsize=9, color=color_text)
        if bbox_fc is not None:
            text_kwargs["bbox"] = dict(facecolor=bbox_fc, alpha=0.35, pad=1.5, edgecolor="none")
        ax.text(row.x_center_nm, y_text, row.contact, **text_kwargs)


def make_bandedges_figure(s, scales):
    fig, axes = plt.subplots(1, 3, figsize=(20, 5.5), constrained_layout=True)

    for ax, Z, title in [
        (axes[0], s["hh_2d"], "Heavy-hole band edge HH(x,z)"),
        (axes[1], s["lh_2d"], "Light-hole band edge LH(x,z)"),
    ]:
        pcm = ax.pcolormesh(
            s["x_be"],
            s["z_be"],
            Z,
            shading="nearest",
            cmap="viridis",
            vmin=scales["band_vmin"],
            vmax=scales["band_vmax"],
        )
        ax.set_xlim(*s["xlim"])
        ax.set_ylim(*s["zlim"])
        ax.set_xlabel("x (nm)")
        ax.set_ylabel("z (nm)")
        ax.set_title(title)
        ax.axhline(0, color="w", ls=":", lw=1, alpha=0.8)
        annotate_gate_lines(ax, s["gate_df"], s["zlim"][1] - 6, "w", "white", bbox_fc="black")
        fig.colorbar(pcm, ax=ax, label="Energy (eV)")

    ax = axes[2]
    ax.plot(s["band_x"]["x[nm]"], s["band_x"]["HH[eV]"], lw=2, label="HH")
    ax.plot(s["band_x"]["x[nm]"], s["band_x"]["LH[eV]"], lw=2, label="LH")
    ax.plot(
        s["band_x"]["x[nm]"],
        s["band_x"]["hole_Fermi_level[eV]"],
        "k--",
        lw=1.8,
        label="hole Fermi level",
    )
    for row in s["gate_df"].itertuples():
        ax.axvline(row.x_center_nm, color="0.5", ls="--", lw=1)
    ax.set_xlim(*s["xlim"])
    ax.set_ylim(*scales["band_line_ylim"])
    ax.set_xlabel("x (nm)")
    ax.set_ylabel("Energy (eV)")
    ax.set_title("Band edges along x at the QW plane")
    ax.grid(alpha=0.2)
    ax.legend(frameon=False)

    fig.suptitle(f"2D band edges | Ge valence bandoffset = {s['offset']:.1f} eV", fontsize=14)
    return fig


def make_density_figure(s):
    fig, axes = plt.subplots(1, 2, figsize=(16, 5.5), constrained_layout=True)

    floor_cm3 = 1e12
    logp2d = np.log10(np.clip(s["p2d_cm3"], floor_cm3, None))

    ax = axes[0]
    pcm = ax.pcolormesh(s["x_den"], s["z_den"], logp2d, shading="nearest", cmap="magma", vmin=12, vmax=19)
    ax.set_xlim(*s["xlim"])
    ax.set_ylim(*s["zlim"])
    ax.set_xlabel("x (nm)")
    ax.set_ylabel("z (nm)")
    ax.set_title("Hole density in x-z plane")
    ax.axhline(0, color="w", ls=":", lw=1, alpha=0.8)
    annotate_gate_lines(ax, s["gate_df"], s["zlim"][1] - 6, "w", "white", bbox_fc="black")
    fig.colorbar(pcm, ax=ax, label=r"$\log_{10}(p\;[\mathrm{cm}^{-3}])$")

    ax = axes[1]
    xcol = s["p1d"].columns[0]
    pcol = s["p1d"].columns[-1]
    p1d_cm3 = np.clip(s["p1d"][pcol].to_numpy() * 1e18, floor_cm3, None)
    ax.semilogy(s["p1d"][xcol], p1d_cm3, lw=2, color="tab:green")
    for row in s["gate_df"].itertuples():
        ax.axvline(row.x_center_nm, color="0.5", ls="--", lw=1)
        ax.text(row.x_center_nm, 2e19, row.contact, ha="center", va="top", fontsize=9, rotation=90)
    ax.set_xlim(*s["xlim"])
    ax.set_ylim(1e12, 2e19)
    ax.set_xlabel("x (nm)")
    ax.set_ylabel(r"$p$ (cm$^{-3}$)")
    ax.set_title("Hole density along x at the QW plane")
    ax.grid(alpha=0.2, which="both")

    fig.suptitle(f"2D hole density | Ge valence bandoffset = {s['offset']:.1f} eV", fontsize=14)
    return fig


def make_vertical_density_figure(s):
    fig, axes = plt.subplots(1, 3, figsize=(17, 4.8), sharey=True, constrained_layout=True)

    floor_cm3 = 1e8

    for ax, (name, df) in zip(axes, s["cuts_density_z"].items()):
        zcol = df.columns[0]
        pcol = df.columns[-1]
        zvals = df[zcol].to_numpy()
        p_cm3_raw = df[pcol].to_numpy() * 1e18
        p_cm3_plot = np.clip(p_cm3_raw, floor_cm3, None)

        ax.semilogy(zvals, p_cm3_plot, lw=2, color="tab:green")
        ax.axvspan(s["z_ge_min"], s["z_ge_max"], color="tab:blue", alpha=0.15)
        ax.axvline(s["z_semi_top"], color="tab:red", ls="--", lw=1.2, alpha=0.8)
        ax.set_xlim(*s["z_window"])
        ax.set_ylim(1e8, 2e19)
        ax.set_xlabel("z (nm)")
        ax.set_title(f"Hole density under {name}")
        ax.grid(alpha=0.2, which="both")

    axes[0].set_ylabel(r"$p$ (cm$^{-3}$)")
    axes[0].text(0.03, 0.95, "Blue band = Ge QW", transform=axes[0].transAxes, va="top")
    axes[0].text(0.03, 0.87, "Red dashed = top semiconductor surface", transform=axes[0].transAxes, va="top")

    fig.suptitle(f"Vertical 2D hole-density cuts | Ge valence bandoffset = {s['offset']:.1f} eV", fontsize=14)
    return fig


def make_potential_figure(s, scales):
    fig, axes = plt.subplots(1, 3, figsize=(20, 5.5), constrained_layout=True)

    ax = axes[0]
    pcm = ax.pcolormesh(
        s["x_phi"],
        s["z_phi"],
        s["phi2d"],
        shading="nearest",
        cmap="viridis",
        vmin=scales["phi_vmin"],
        vmax=scales["phi_vmax"],
    )
    ax.set_title("Potential: full 2D x-z map")
    ax.set_xlabel("x (nm)")
    ax.set_ylabel("z (nm)")
    rect = Rectangle(
        (s["x_window"][0], s["z_window"][0]),
        s["x_window"][1] - s["x_window"][0],
        s["z_window"][1] - s["z_window"][0],
        fill=False,
        ec="white",
        lw=1.5,
        ls="--",
    )
    ax.add_patch(rect)
    ax.text(
        s["x_window"][0] + 10,
        s["z_window"][1] - 8,
        "active window for line cuts",
        color="white",
        fontsize=10,
        ha="left",
        va="top",
        bbox=dict(facecolor="black", alpha=0.3, pad=2, edgecolor="none"),
    )
    fig.colorbar(pcm, ax=ax, label="Potential (V)")

    ax = axes[1]
    xcol = s["phi_x"].columns[0]
    ycol = s["phi_x"].columns[-1]
    ax.plot(s["phi_x"][xcol], s["phi_x"][ycol], lw=2, color="black")
    for row in s["gate_df"].itertuples():
        ax.axvline(row.x_center_nm, color="0.5", ls="--", lw=1)
    ax.set_xlim(*s["x_window"])
    ax.set_ylim(*scales["phi_line_ylim"])
    ax.set_xlabel("x (nm)")
    ax.set_ylabel("Potential (V)")
    ax.set_title("Potential along x at the QW plane")
    ax.grid(alpha=0.2)

    ax = axes[2]
    cut_colors = {"PG1": "tab:red", "BG2": "tab:cyan", "PG2": "tab:orange"}
    for name, df in s["cuts_phi_z"].items():
        zcol = df.columns[0]
        ycol = df.columns[-1]
        ax.plot(df[zcol], df[ycol], lw=2, label=name, color=cut_colors[name])
    ax.axvspan(s["z_ge_min"], s["z_ge_max"], color="tab:blue", alpha=0.15, label="Ge QW")
    ax.axvline(s["z_semi_top"], color="tab:red", ls="--", lw=1.2, label="Top semiconductor surface")
    ax.set_xlim(*s["z_window"])
    ax.set_ylim(*scales["phi_line_ylim"])
    ax.set_xlabel("z (nm)")
    ax.set_ylabel("Potential (V)")
    ax.set_title("Potential along vertical cuts")
    ax.grid(alpha=0.2)
    ax.legend(frameon=False)

    fig.suptitle(f"2D potential | Ge valence bandoffset = {s['offset']:.1f} eV", fontsize=14)
    return fig


original_offset = get_ge_valence_bandoffset(DATABASE_NNP)
print(f"Original Ge valence-band bandoffset: {original_offset:.15g} eV")
print(f"Sweeping {len(BANDOFFSET_VALUES)} values: {BANDOFFSET_VALUES[0]:.1f} to {BANDOFFSET_VALUES[-1]:.1f} eV")
print(f"GIF output folder: {GIF_DIR}")

snapshots = []
backup_created = False
sweep_start = time.perf_counter()

try:
    for i, offset in enumerate(BANDOFFSET_VALUES, start=1):
        set_ge_valence_bandoffset(float(offset), database_path=DATABASE_NNP, make_backup=not backup_created)
        backup_created = True

        t0 = time.perf_counter()
        inp = nna.run_input(INPUT_DOUBLE_DOT_2D)
        run_root = Path(inp.folder_output)
        snapshots.append(load_2d_snapshot(run_root, offset))
        dt = time.perf_counter() - t0

        print(
            f"[{i:02d}/{len(BANDOFFSET_VALUES):02d}] "
            f"bandoffset = {offset:.1f} eV | run time = {dt:5.1f} s | output = {run_root}"
        )
finally:
    set_ge_valence_bandoffset(original_offset, database_path=DATABASE_NNP, make_backup=False)
    print(f"Restored Ge valence-band bandoffset to {original_offset:.15g} eV")

if not snapshots:
    raise RuntimeError("No successful 2D runs completed, so no GIFs were created.")

scales = compute_scales(snapshots)

gif_paths = {
    "bandedges": GIF_DIR / "2d_bandedges_vs_ge_bandoffset.gif",
    "density": GIF_DIR / "2d_hole_density_vs_ge_bandoffset.gif",
    "vertical_density": GIF_DIR / "2d_vertical_hole_cuts_vs_ge_bandoffset.gif",
    "potential": GIF_DIR / "2d_potential_vs_ge_bandoffset.gif",
}

frames = [fig_to_pil(make_bandedges_figure(s, scales)) for s in snapshots]
save_gif(frames, gif_paths["bandedges"])
del frames

frames = [fig_to_pil(make_density_figure(s)) for s in snapshots]
save_gif(frames, gif_paths["density"])
del frames

frames = [fig_to_pil(make_vertical_density_figure(s)) for s in snapshots]
save_gif(frames, gif_paths["vertical_density"])
del frames

frames = [fig_to_pil(make_potential_figure(s, scales)) for s in snapshots]
save_gif(frames, gif_paths["potential"])
del frames

elapsed = time.perf_counter() - sweep_start
print(f"Finished sweep and GIF export in {elapsed/60:.1f} minutes")
for name, path in gif_paths.items():
    print(f"{name:>16}: {path}")
    if SHOW_INLINE:
        display(IPyImage(filename=str(path)))
```
